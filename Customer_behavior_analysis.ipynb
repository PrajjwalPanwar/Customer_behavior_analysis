### Handling Missing Values

The initial dataset contains 37 missing values in the Review Rating column.

We selected median imputation because the median is less sensitive to outliers than the mean.

Instead of calculating one overall median, we calculated the median review rating separately for each product category.

The missing values were replaced using their corresponding category median.

```python
# Cell 6
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# Cell 7
df.isnull().sum()
```

```python
# Cell 8
# use snake-case to represent column name 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
# Cell 9
df.columns
```

```python
# Cell 10
# Create a column age_group 
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
# Cell 11
df[['age','age_group']].head(10)
```

```python
# Cell 12
# create column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
# Cell 16
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1)
```

```python
# Cell 18
from sqlalchemy import create_engine, URL
from getpass import getpass

password = getpass("Enter your MySQL password: ")

connection_url = URL.create(
    drivername="mysql+mysqlconnector",
    username="root",
    password=password,
    host="localhost",
    port=3306,
    database="customer_shopping_db"
)

engine = create_engine(connection_url)

with engine.connect() as connection:
    print("Connected successfully!")
    print(
        "Current database:",
        connection.exec_driver_sql("SELECT DATABASE()").scalar()
)
```

```python
# Cell 19
# Export the cleaned dataset to MySQL

print("DataFrame shape before export:", df.shape)

df.to_sql(
    name="customer_shopping_behavior_cleaned",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
# Cell 20
# Verify the cleaned dataset in MySQL

mysql_sample = pd.read_sql(
    "SELECT * FROM customer_shopping_behavior_cleaned LIMIT 10",
    con=engine
)

mysql_sample.head()
```

```python
# Cell 21
# Verify total records stored in MySQL

row_count = pd.read_sql(
    """
    SELECT COUNT(*) AS total_records
    FROM customer_shopping_behavior_cleaned
    """,
    con=engine
)

row_count
```
